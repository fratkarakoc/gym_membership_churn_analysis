| Column | Description |
|---|---|
| `gender` | Member's gender, binary encoded (0/1) |
| `Near_Location` | Whether the member lives or works near the gym (1 = yes, 0 = no) |
| `Partner` | Whether the member is affiliated with a partner company offering a corporate gym discount |
| `Promo_friends` | Whether the member originally signed up through a "bring a friend" promo code |
| `Phone` | Whether the member provided a contact phone number |
| `Contract_period` | Length of the current membership contract, in months (1, 6, or 12) |
| `Group_visits` | Whether the member takes part in group classes |
| `Age` | Member's age, in years |
| `Avg_additional_charges_total` | Total spending on additional gym services (cafe, personal training, sports goods, etc.) |
| `Month_to_end_contract` | Months remaining until the current contract expires |
| `Lifetime` | Months since the member's first sign-up at the gym |
| `Avg_class_frequency_total` | Average weekly visit frequency, over the member's entire membership lifetime |
| `Avg_class_frequency_current_month` | Average weekly visit frequency, current month only |
| `Churn` | Target variable — whether the member churned (left) the following month (1 = churned, 0 = stayed) |
