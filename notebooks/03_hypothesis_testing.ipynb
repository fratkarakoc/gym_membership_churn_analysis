```python
import pandas as pd
import numpy as np
from scipy.stats import mannwhitneyu, chi2_contingency
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

sns.set()
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv('../data/gym_churn_us_clean.csv')
data.head()
```

Testing if the frequency drop is higher on churn positive members  
&darr; &darr; &darr;

```python
churn_positive = data.loc[data['churn'] == 1, 'frequency_drop']
churn_negative = data.loc[data['churn'] == 0, 'frequency_drop']

u, p_value = mannwhitneyu(churn_negative, churn_positive, alternative='less')

print(f"U = {u}")
print(f"p value = {p_value}")
```

&uarr; &uarr; &uarr;  
We used the mann-whitney u test because frequency drops' values on both groups are not normal distributed

##### Categorical Tests

Gender

```python
gender_table = pd.crosstab(data['gender'], data['churn'])

chi2, pvalue, df, expected = chi2_contingency(gender_table)

n = gender_table.to_numpy().sum()
r, c = gender_table.shape
gender_cramers_v = np.sqrt(chi2 / (n * min(r-1,c-1)))

print(f"Chi2: {chi2}")
print(f"P-value: {pvalue}")
print(f"Degrees of Freedom: {df}")
print(f"Gender Table:")
print(f"{gender_table}")
print(f"Expected:")
print(f"{expected}")
print(f"Cramer's V:")
print(f"{gender_cramers_v}")
```
